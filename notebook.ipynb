## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** "Bagaimana kecenderungan tingkat penyewaan sepeda oleh pengguna kategori kasual maupun terdaftar apabila dianalisis berdasarkan pengaruh musim serta kondisi cuaca selama tahun 2011 sampai 2012?"
- **Pertanyaan 2:** "Bagaimana perbandingan pola waktu operasional (jam dalam sehari) antara hari kerja dengan akhir pekan terhadap jumlah penyewaan, serta seberapa erat hubungan linear antara suhu udara dengan total penyewaan harian?"

## Exploratory Data Analysis (EDA)

### Explore data bike sharing

> Pada tahap *Exploratory Data Analysis (EDA)* ini, saya melakukan penggalian informasi secara terarah untuk menjawab dua pertanyaan bisnis utama yang telah saya tentukan di awal. Berikut informasi yang saya gunakan:

>> * **Analisis Pengaruh Musim dan Cuaca (Fokus Pertanyaan Bisnis 1)**: Saya mengelompokkan dan menganalisis data untuk melihat bagaimana perubahan musim (*season*) dan kondisi cuaca (*weathersit*) memengaruhi fluktuasi jumlah penyewaan sepeda harian (`cnt`).
>> * **Analisis Perbandingan Hari Kerja vs Akhir Pekan (Fokus Pertanyaan Bisnis 2)**: Saya membedah pola penyewaan berdasarkan kategori hari kerja (*workingday*) versus akhir pekan atau hari libur untuk memahami perbedaan perilaku pengguna, volume transaksi, serta korelasinya dengan faktor suhu.
>> * **Ringkasan Statistik Dasar**: Saya menggunakan fungsi statistik deskriptif untuk melihat distribusi data, nilai rata-rata, minimum, serta maksimum dari jumlah penyewaan sebagai fondasi angka yang objektif sebelum divisualisasikan.

> Melalui proses eksplorasi ini, saya dapat mengumpulkan hasil yang nantinya akan saya berikan dalam bentuk visualisasi untuk menjawab kedua pertanyaan bisnis secara baik dan akurat.

```python
season_mapping = {1: 'Springer', 2: 'Summer', 3: 'Fall', 4: 'Winter'}
weathersit_mapping = {1: 'Clear/Partly Cloudy', 2: 'Misty/Cloudy', 3: 'Light Snow/Rain', 4: 'Severe Weather'}

rental_harian_df['season_label'] = rental_harian_df['season'].map(season_mapping)
rental_harian_df['weathersit_label'] = rental_harian_df['weathersit'].map(weathersit_mapping)

print("Kolom label musim dan cuaca berhasil dibuat!")
```

```python
analisis_musim_df = rental_harian_df.groupby('season_label').agg({
    'casual': 'mean',
    'registered': 'mean',
    'cnt': 'mean'
}).reset_index()

print("--- RATA-RATA PENYEWAAN BERDASARKAN MUSIM ---")
display(analisis_musim_df)

analisis_cuaca_df = rental_harian_df.groupby('weathersit_label').agg({
    'casual': 'mean',
    'registered': 'mean',
    'cnt': 'mean'
}).reset_index()

print("\n--- RATA-RATA PENYEWAAN BERDASARKAN KONDISI CUACA ---")
display(analisis_cuaca_df)
```

```python
pola_jam_df = rental_jam_df.groupby(['workingday', 'hr'])['cnt'].mean().reset_index()
print("--- RATA-RATA PENYEWAAN PER JAM (WORKINGDAY VS LIBUR) ---")
display(pola_jam_df.head(10))

korelasi_suhu = rental_harian_df[['temp', 'cnt']].corr()
print("\n--- NILAI KORELASI SUHU TERHADAP PENYEWAAN HARIAN ---")
display(korelasi_suhu)
```

```python
print("--- STATISTIK DESKRIPTIF (DESCRIBE) ---")
display(rental_harian_df[['cnt', 'casual', 'registered', 'temp', 'hum']].describe())

plt.figure(figsize=(14, 4))
plt.subplot(1, 3, 1)
sns.histplot(rental_harian_df['cnt'], kde=True, color='skyblue')
plt.title('Distribusi Total Penyewaan (cnt)', fontsize=10, fontweight='bold')

plt.subplot(1, 3, 2)
sns.histplot(rental_harian_df['casual'], kde=True, color='salmon', label='Casual')
sns.histplot(rental_harian_df['registered'], kde=True, color='seagreen', label='Registered')
plt.title('Distribusi Casual vs Registered', fontsize=10, fontweight='bold')
plt.legend()

plt.subplot(1, 3, 3)
sns.boxplot(data=rental_harian_df[['cnt', 'casual', 'registered']], palette='Set2')
plt.title('Boxplot Deteksi Outlier', fontsize=10, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
numeric_cols = ['temp', 'atemp', 'hum', 'windspeed', 'casual', 'registered', 'cnt']
correlation_matrix = rental_harian_df[numeric_cols].corr()

sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Matrix Antar Variabel Numerik', fontsize=11, fontweight='bold')
plt.tight_layout()
plt.show()
```

**Insight:** (Opsional)
- **Eksplorasi Bagian 1 (Musim dan Cuaca):**
  > Berdasarkan hasil agregasi data, kategori musim gugur (Fall) mencatatkan rata-rata volume penyewaan tertinggi dibandingkan musim lainnya, dengan dominasi yang sangat kuat dari pengguna terdaftar (registered). Selain itu, kondisi cuaca cerah (Clear/Partly Cloudy) memberikan tingkat kenyamanan optimal yang menghasilkan rata-rata transaksi penyewaan tertinggi secara keseluruhan.
- **Eksplorasi Bagian 2 (Pola Waktu dan Korelasi Suhu):**
  > Analisis statistik per jam menunjukkan adanya pola jam sibuk (rush hour) yang spesifik pada hari kerja, sedangkan uji korelasi antara suhu udara (temp) dan total penyewaan harian menunjukkan angka positif (korelasi sedang hingga kuat), yang mengindikasikan bahwa peningkatan suhu udara sejalan dengan naiknya minat masyarakat menggunakan sepeda.

- **(Univariate & Multivariate):**
 > Berdasarkan hasil statistik deskriptif (describe), rata-rata total penyewaan harian (cnt) adalah sekitar 4.504 unit. Grafik histogram menunjukkan pola distribusi data yang cukup seimbang, sementara boxplot mendeteksi beberapa nilai ekstrem (outlier) pada kelompok pengguna kasual (casual). Selain itu, berdasarkan matriks korelasi (heatmap), variabel suhu (temp dan atemp) memiliki korelasi positif yang kuat terhadap total penyewaan sepeda. ## Analisis Lanjutan (Analisis RFM (Recency, Frequency, Monetary))

```python
analisis_hari_df = rental_harian_df.groupby('weekday').agg({
    'casual': 'mean',
    'registered': 'mean',
    'cnt': 'mean'
}).reset_index()

mapping_hari = {0: 'Minggu', 1: 'Senin', 2: 'Selasa', 3: 'Rabu', 4: 'Kamis', 5: 'Jumat', 6: 'Sabtu'}
analisis_hari_df['nama_hari'] = analisis_hari_df['weekday'].map(mapping_hari)

print("--- RATA-RATA PENYEWAAN BERDASARKAN HARI DALAM SEMINGGU ---")
display(analisis_hari_df[['nama_hari', 'casual', 'registered', 'cnt']])

plt.figure(figsize=(10, 4))
sns.barplot(data=analisis_hari_df, x='nama_hari', y='cnt', hue='nama_hari', palette='viridis', legend=False, errorbar=None)
plt.title('Rata-rata Total Penyewaan Sepeda Berdasarkan Hari', fontsize=11, fontweight='bold')
plt.xlabel('Hari dalam Seminggu', fontsize=10)
plt.ylabel('Rata-rata Total Penyewaan', fontsize=10)
plt.tight_layout()
plt.show()
```

```python
analisis_rasio_df = rental_harian_df.groupby('workingday').agg({
    'casual': 'mean',
    'registered': 'mean',
    'cnt': 'mean'
}).reset_index()

analisis_rasio_df['kategori_hari'] = analisis_rasio_df['workingday'].map({0: 'Akhir Pekan / Libur', 1: 'Hari Kerja'})

print("--- RATA-RATA TRANSAKSI BERDASARKAN HARI KERJA VS LIBUR ---")
display(analisis_rasio_df[['kategori_hari', 'casual', 'registered', 'cnt']])

plt.figure(figsize=(9, 4))
analisis_rasio_melted = analisis_rasio_df.melt(id_vars=['kategori_hari'], value_vars=['casual', 'registered'],
                                               var_name='Tipe Pengguna', value_name='Rata-rata Penyewaan')

sns.barplot(data=analisis_rasio_melted, x='kategori_hari', y='Rata-rata Penyewaan', hue='Tipe Pengguna', palette='Set2')
plt.title('Perbandingan Pengguna Kasual vs Terdaftar (Hari Kerja vs Libur)', fontsize=11, fontweight='bold')
plt.xlabel('Kategori Hari', fontsize=10)
plt.ylabel('Rata-rata Volume Penyewaan', fontsize=10)
plt.legend(title='Tipe Pengguna')
plt.tight_layout()
plt.show()
```

```python
def categorize_rentals(x):
    if x < 3000:
        return 'Rendah (< 3k)'
    elif x <= 6000:
        return 'Sedang (3k - 6k)'
    else:
        return 'Tinggi (> 6k)'

rental_harian_df['rental_category'] = rental_harian_df['cnt'].apply(categorize_rentals)

category_counts = rental_harian_df['rental_category'].value_counts().reset_index()
category_counts.columns = ['kategori', 'jumlah_hari']

order_category = ['Rendah (< 3k)', 'Sedang (3k - 6k)', 'Tinggi (> 6k)']

plt.figure(figsize=(8, 4))
sns.barplot(data=category_counts, x='kategori', y='jumlah_hari', order=order_category, palette='Blues_r', hue='kategori', legend=False)
plt.title('Distribusi Frekuensi Kategori Volume Penyewaan Harian', fontsize=11, fontweight='bold')
plt.xlabel('Kategori Volume Penyewaan', fontsize=10)
plt.ylabel('Jumlah Hari', fontsize=10)
plt.tight_layout()
plt.show()

print("Ringkasan Jumlah Hari Berdasarkan Kategori Penyewaan:")
display(category_counts)
```
